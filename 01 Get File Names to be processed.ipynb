```python
import glob
```

Glob is a Python module (import glob) used for pathname pattern expansion, allowing you to find files and directories that match Unix shell-style wildcards. It supports patterns like * (any characters), ? (single character), and [seq] (character sequences).

```python
glob.glob('data/retail_db/**', recursive=True)
```

```python
glob.glob('data/retail_db/*/*')
```

```python
src_file_names = glob.glob('data/retail_db/*/part-*')
```

```python
src_file_names
```

```python
import pandas as pd
```

```python
for file_name in src_file_names:
    df = pd.read_csv(file_name, header=None)
    print(f'Shape of {file_name} is {df.shape}')
```
